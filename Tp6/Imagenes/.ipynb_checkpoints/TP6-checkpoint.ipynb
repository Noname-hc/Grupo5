## Ejercicios Teóricos

1. ¿En qué tipo de algoritmos se basa un planificador para encontrar el mejor camino a un estado solución?

Un __planificador__ se fundamenta en algoritmos de *búsqueda en grafos* o en *espacios de estados*. Estos algoritmos permiten explorar diferentes secuencias de acciones hasta alcanzar un estado meta que satisfaga las condiciones deseadas.

Entre los más relevantes se encuentran:

- __Búsquedas no informadas (o ciegas):__ como la búsqueda en amplitud (BFS), en profundidad (DFS) o la búsqueda en profundidad iterativa. Estas no emplean conocimiento adicional, por lo que suelen ser menos eficientes.

- __Búsquedas informadas (heurísticas):__ como A* o Greedy Best-First Search, que utilizan funciones heurísticas para guiar la exploración de estados hacia la solución más prometedora.

- __Algoritmos especializados de planificación:__ como Graphplan o SATPlan, que representan de manera compacta el espacio de estados y aplican técnicas de optimización.

En conclusión, los planificadores suelen recurrir principalmente a algoritmos heurísticos ya que permiten encontrar soluciones en menos tiempo y con menor costo computacional.



2. ¿Qué tres elementos se encuentran dentro de una acción formulada en lenguaje STRIPS? Describa brevemente qué función cumple cada uno.

En el modelo de planificación __STRIPS__ (Stanford Research Institute Problem Solver), cada acción está definida a través de tres componentes esenciales:

- __Precondiciones:__ Son los hechos o condiciones que deben cumplirse en el estado actual para que la acción pueda ejecutarse. Ejemplo: para realizar la acción abrir puerta, es necesario tener la llave.

- __Efectos positivos (add list):__ Representan los hechos que pasan a ser verdaderos una vez ejecutada la acción. Ejemplo: tras abrir la puerta, el hecho puerta abierta se agrega al estado.

- __Efectos negativos (delete list):__ Indican los hechos que dejan de cumplirse al ejecutar la acción. Ejemplo: al abrir la puerta, el hecho puerta cerrada se elimina del estado.

Este esquema permite modelar de forma sencilla y estructurada los cambios que las acciones producen en el mundo.

3. Describa las ventajas y desventajas de desarrollar un algoritmo de planificación hacia adelante y hacia atrás en el espacio de estados.


Existen dos enfoques principales en la planificación clásica: la *planificación hacia adelante* (forward search) y la *planificación hacia atrás* (backward search). Ambos presentan ventajas y limitaciones que deben considerarse según el problema a resolver.

### Planificación hacia adelante:

- __Ventajas:__

El proceso inicia desde el estado inicial, lo que facilita la aplicación práctica y la consistencia con el mundo real.

Solo se generan acciones que son ejecutables en el contexto actual.

- __Desventajas:__

Puede explorar una gran cantidad de estados irrelevantes antes de alcanzar la meta.

Requiere de buenas heurísticas para ser eficiente, de lo contrario el proceso es costoso.

### Planificación hacia atrás:

- __Ventajas:__

El proceso comienza desde el objetivo, lo que reduce la exploración de estados innecesarios.

Es más eficiente cuando los objetivos están claramente definidos y son pocos.

- __Desventajas:__

Puede ser difícil encontrar las acciones que satisfagan las condiciones del objetivo.

Si las metas poseen múltiples precondiciones generales, el proceso se vuelve complejo y costoso.
